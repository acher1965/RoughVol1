# RoughVol1 Walkthrough

This notebook shows how to check inputs, run a single request, and run checks. It is safe to execute without the input spreadsheet; it will skip the run if the file is missing.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd()
print(f"Project root: {PROJECT_ROOT}")
```

## Input file expectations
- Default input file: `RFSV_input.xlsx` (override with `-x`).
- Required sheets: `VolGrid` and `WeightsGrid`.
- Main sheet columns include `Destination`, `Row`, `Date`, `Spot`, `Underlying`, `Simulations`, `Seed`, `Roughness(Hurst)`, `VolVol(Eta)`, `Correlation(Rho)`, `FullDiagnostic`, `FullOutput`, `Tenor1..TenorN`, `Fwd1..FwdN`, `Impvar1..ImpvarN`.

```python
input_path = PROJECT_ROOT / "RFSV_input.xlsx"
if input_path.exists():
    print(f"Found input file: {input_path}")
else:
    print(f"Input file not found: {input_path}")
    print("Place the file here or pass -x with a different path.")
```

```python
# Guarded example: only show the command if the file exists.
if input_path.exists():
    cmd = [sys.executable, "RoughVol1/RFSV_main.py", "-x", str(input_path), "1"]
    print("You can run a single request with:")
    print(" ".join(cmd))
else:
    print("Skipping run example because the input file is missing.")
```
